```python
#Rewriting the above code using a for loop

# URLs of the images
image_urls = [
    'https://upload.wikimedia.org/wikipedia/en/c/cb/Monkey_D_Luffy.png',
    'https://static.wikia.nocookie.net/p__/images/6/6e/Roronoa_Zoro_Post_Timeskip.png/revision/latest?cb=20190529044836&path-prefix=protagonist',
    'https://static.wikia.nocookie.net/p__/images/f/f0/Nami_Waifuuu.png/revision/latest?cb=20190113023031&path-prefix=protagonist',
    'https://static.wikia.nocookie.net/p__/images/5/53/Usopp.png/revision/latest?cb=20190529045324&path-prefix=protagonist',
    'https://static.wikia.nocookie.net/p__/images/b/b0/Sanji_Vector.png/revision/latest?cb=20190426102221&path-prefix=protagonist'
]

# Download and open images
images = []
for url in image_urls:
    response = requests.get(url)
    image_file = BytesIO(response.content)
    img = Image.open(image_file)
    images.append(img)

# Coordinates and dimensions for each image
axes_coords = [
    [0.05, 0.55, 0.15, 0.35],  # Luffy
    [0.25, 0.2, 0.15, 0.35],   # Zoro
    [0.45, 0.12, 0.15, 0.35],   # Nami
    [0.65, 0.35, 0.15, 0.35],   # Usopp
    [0.87, 0.11, 0.15, 0.35]    # Sanji
]

# Create the plot
fig, ax = plt.subplots(figsize=(7, 3), layout="constrained")
ax.plot(x, y2, color="red", alpha=0.5, linestyle='--')

# Add images to the plot
for i, (img, coords) in enumerate(zip(images, axes_coords)):
    ax_img = fig.add_axes(coords)  # Create axes for the image
    ax_img.imshow(img)             # Display the image
    ax_img.axis('off')             # Hide axes

# Add labels and titles
ax.set_xlabel('Characters', fontsize=14)
ax.set_ylabel('Power', fontsize=14)
ax.set_title('Power Levels of Pirate Crew', fontsize=18)

# Show the plot
plt.show()
```
